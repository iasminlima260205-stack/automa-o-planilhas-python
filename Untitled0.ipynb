```python
vendas =[100,80,120]
total = sum(vendas)

print("===RELATÓRIO DE VENDAS===")
print(F"Total vendido: R$ {total}")
print(f"Total de vendas: {len(vendas)}")
```

```python
import matplotlib.pyplot as plt

vendas = [100,80,120,200,150]
dias = ["Seg", "Ter", "Qua", "Qui", "Sex"]
total = sum(vendas)
media = total / len(vendas)
maior = max(vendas)

print(f"Total:R$ {total}")
print(f"Média por dia: R$ {media:.2f}")
print(f"Maior venda: R$ {maior}")

plt.bar(dias, vendas)
plt.title("Vendas da Semana")
plt.ylabel("R$ Vendido")
plt.show()
```
